```python
print(banking.isnull().sum())
banking["Customer_ID"] = banking["Customer_ID"].fillna("Unknown")
banking["Customer_Name"] = banking["Customer_Name"].fillna("Unknown")
banking["State"] = banking["State"].fillna("Unknown")
banking["Merchant_ID"] = banking["Merchant_ID"].fillna("Unknown")
banking["Transaction_Device"] = banking["Transaction_Device"].fillna("Unknown")
banking["Transaction_Location"] = banking["Transaction_Location"].fillna("Unknown")
banking["Customer_Contact"] = banking["Customer_Contact"].fillna("Unknown")
banking["Transaction_Description"] = banking["Transaction_Description"].fillna("Unknown")
banking["Customer_Email"] = banking["Customer_Email"].fillna("Unknown")
```
