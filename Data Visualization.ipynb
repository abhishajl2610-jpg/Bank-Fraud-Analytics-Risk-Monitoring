```python
import pandas as pd;
import numpy as np;
import matplotlib.pyplot as plt;
```

```python
from pathlib import Path
BASE_DIR = Path.cwd()
```

```python
banking = pd.read_excel(
    BASE_DIR / "Bank_Fraud_Feature_Engineered (Final).xlsx"
)

print("Dataset Loaded Successfully")
```

```python
print(banking.columns)
```

```python
# What is the distribution of fraud and non-fraud transactions?

fraud=banking[banking["Is_Fraud"]==1];
fraud_counts=banking["Is_Fraud"].value_counts()
print(fraud_counts)

plt.figure(figsize=(7, 2))

plt.barh(
    ["Non-Fraud", "Fraud"],
    [
        fraud_counts[0],
        fraud_counts[1]
    ]
)

plt.title("Fraud vs Non-Fraud Transactions")
plt.xlabel("Transaction Category")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
# Which transaction type has the highest number of fraudulent transactions?fraud=banking[banking["Is_Fraud"]==1]

fraud_by_type=fraud["Transaction_Type"].value_counts()
print(fraud_by_type)

plt.figure(figsize=(7,5))

plt.bar(fraud_by_type.index,
    fraud_by_type.values)

plt.title("Fraudulent Transactions by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Fraudulent Transactions")

# plt.xticks(rotation=45)

plt.show()
```

```python
# Which merchant category has the highest number of fraudulent transactions?
fraud_by_merchant=fraud["Merchant_Category"].value_counts();
print(fraud_by_merchant)

plt.figure(figsize=(9,5))
plt.bar(fraud_by_merchant.index,fraud_by_merchant.values)

plt.title("Fraudulent Transactions by Merchant Category")
plt.xlabel("Merchant Category")
plt.ylabel("Number of Fraudulent Transactions")

plt.xticks(rotation=30)
plt.show()
```

```python
# Which device type is most frequently used for fraudulent transactions?

fraud_by_device = fraud["Device_Type"].value_counts()
print(fraud_by_device)

plt.figure(figsize=(8, 5))
bars = plt.bar(
    fraud_by_device.index,
    fraud_by_device.values
)

plt.title("Fraudulent Transactions by Device Type")
plt.xlabel("Device Type")
plt.ylabel("Number of Fraudulent Transactions")

# Show counts
for bar in bars:
    count = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        count,
        int(count),
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
# At what time of day do fraudulent transactions occur most frequently?

fraud_by_time = fraud["Time_of_Day"].value_counts()
print(fraud_by_time)

plt.figure(figsize=(8, 5))
bars = plt.bar(
    fraud_by_time.index,
    fraud_by_time.values
)
plt.title("Fraudulent Transactions by Time of Day")
plt.xlabel("Time of Day")
plt.ylabel("Number of Fraudulent Transactions")

# Show counts
for bar in bars:
    count = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        count,
        int(count),
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
# How are fraudulent transactions distributed across Low, Medium, and High Risk levels?

fraud_by_risk = fraud["Risk_Level"].value_counts()
print(fraud_by_risk)

plt.figure(figsize=(8, 5))
bars = plt.bar(
    fraud_by_risk.index,
    fraud_by_risk.values
)
plt.title("Fraudulent Transactions by Risk Level")
plt.xlabel("Risk Level")
plt.ylabel("Number of Fraudulent Transactions")

# Show counts
for bar in bars:
    count = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        count,
        int(count),
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
# Which age group has the highest number of fraudulent transactions?

fraud_by_age = fraud["Age_Group"].value_counts()
print(fraud_by_age)

plt.figure(figsize=(8, 5))
bars = plt.bar(
    fraud_by_age.index,
    fraud_by_age.values
)
plt.title("Fraudulent Transactions by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Fraudulent Transactions")

# Show counts
for bar in bars:
    count = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        count,
        int(count),
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
# Which transaction size category has the highest number of fraudulent transactions?

fraud_by_size = fraud["Transaction_Size"].value_counts()
print(fraud_by_size)

plt.figure(figsize=(8, 5))
bars = plt.bar(
    fraud_by_size.index,
    fraud_by_size.values
)
plt.title("Fraudulent Transactions by Transaction Size")
plt.xlabel("Transaction Size")
plt.ylabel("Number of Fraudulent Transactions")
plt.show()
```

```python
# Top 10 states with the highest number of fraudulent transactions.

fraud_by_state = fraud["State"].value_counts().head(10)
print(fraud_by_state)

plt.figure(figsize=(9, 6))
bars = plt.barh(
    fraud_by_state.index,
    fraud_by_state.values
)

plt.title("Top 10 States by Number of Fraudulent Transactions")
plt.xlabel("Number of Fraudulent Transactions")
plt.ylabel("State")

# Show counts
for bar in bars:
    count = bar.get_width()
    plt.text(
        count,
        bar.get_y() + bar.get_height() / 2,
        int(count),
        ha="left",
        va="center"
    )

# Highest state at the top
plt.gca().invert_yaxis()

plt.show()
```

```python
# How are transaction amounts distributed across the dataset?

plt.figure(figsize=(9, 5))
plt.hist(
    banking["Transaction_Amount"],
    bins=30
)

plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
# Fraud Rate by Transaction Type

fraud_rate_transaction = (
    banking.groupby("Transaction_Type")["Is_Fraud"]
    .mean() * 100
)
print(fraud_rate_transaction.sort_values(ascending=False))
plt.figure(figsize=(9, 5))
bars = plt.bar(
    fraud_rate_transaction.index,
    fraud_rate_transaction.values
)
plt.title("Fraud Rate by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Fraud Rate (%)")

# Show percentage on each bar
for bar in bars:
    rate = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        rate,
        f"{rate:.2f}%",
        ha="center",
        va="bottom"
    )

plt.xticks(rotation=30)

plt.show()

highest_fraud_rate_transaction = fraud_rate_transaction.idxmax()
highest_fraud_rate_value = fraud_rate_transaction.max()

print(
    f"Transaction type with the highest fraud rate: "
    f"{highest_fraud_rate_transaction} "
    f"({highest_fraud_rate_value:.2f}%)"
)
```

```python
# Transaction Amount: Fraud vs Non-Fraud
# Do fraudulent transactions have a different transaction-amount distribution compared with non-fraudulent transactions?

plt.figure(figsize=(8, 5))

plt.boxplot(
    [
        banking[banking["Is_Fraud"] == 0]["Transaction_Amount"],
        banking[banking["Is_Fraud"] == 1]["Transaction_Amount"]
    ],
    tick_labels=["Non-Fraud", "Fraud"]
)

plt.title("Transaction Amount: Fraud vs Non-Fraud")
plt.xlabel("Fraud Status")
plt.ylabel("Transaction Amount")

plt.show()
```

```python
# Account Balance vs Transaction Amount

plt.figure(figsize=(9, 6))

plt.scatter(
    banking[banking["Is_Fraud"] == 0]["Account_Balance"],
    banking[banking["Is_Fraud"] == 0]["Transaction_Amount"],
    alpha=0.3,
    label="Non-Fraud"
)

plt.scatter(
    banking[banking["Is_Fraud"] == 1]["Account_Balance"],
    banking[banking["Is_Fraud"] == 1]["Transaction_Amount"],
    alpha=0.5,
    label="Fraud"
)

plt.title("Account Balance vs Transaction Amount")
plt.xlabel("Account Balance")
plt.ylabel("Transaction Amount")

plt.legend()

plt.show()
```

```python
# Balance Utilization: Fraud vs Non-Fraud

plt.figure(figsize=(8, 5))

plt.boxplot(
    [
        banking[banking["Is_Fraud"] == 0]["Balance_Utilization"],
        banking[banking["Is_Fraud"] == 1]["Balance_Utilization"]
    ],
    tick_labels=["Non-Fraud", "Fraud"]
)

plt.title("Balance Utilization: Fraud vs Non-Fraud")
plt.xlabel("Fraud Status")
plt.ylabel("Balance Utilization")

plt.show()
```

```python
# Risk Score vs Fraud

risk_by_fraud = banking.groupby("Is_Fraud")["Risk_Score"].mean()
print(risk_by_fraud)

plt.figure(figsize=(4, 5))
bars = plt.bar(
    ["Non-Fraud", "Fraud"],
    risk_by_fraud.values
)

plt.title("Average Risk Score: Fraud vs Non-Fraud")
plt.xlabel("Fraud Status")
plt.ylabel("Average Risk Score")

for bar in bars:
    score = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        score,
        f"{score:.2f}",
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
# Fraud Rate by Risk Score

fraud_rate_score = (
    banking.groupby("Risk_Score")["Is_Fraud"]
    .mean() * 100
)
print(fraud_rate_score)

plt.figure(figsize=(10, 5))

plt.plot(
    fraud_rate_score.index,
    fraud_rate_score.values,
    marker="o"
)

plt.title("Fraud Rate by Risk Score")
plt.xlabel("Risk Score")
plt.ylabel("Fraud Rate (%)")

plt.grid(True, alpha=0.3)

plt.show()
```

```python
# Correlation with Fraud

correlation = banking[
    [
        "Age",
        "Transaction_Amount",
        "Account_Balance",
        "Balance_Utilization",
        "Risk_Score",
        "Is_Fraud"
    ]
].corr()

print(correlation)

plt.figure(figsize=(9, 7))

plt.imshow(correlation, cmap="coolwarm")

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Matrix of Fraud-Related Variables")

plt.show()
```
