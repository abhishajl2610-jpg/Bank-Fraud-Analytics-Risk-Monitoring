```python
# Outlier Analysis

# Transaction Amount

Q1 = banking["Transaction_Amount"].quantile(0.25)
Q3 = banking["Transaction_Amount"].quantile(0.75)
IQR = Q3 - Q1

outliers = banking[
    (banking["Transaction_Amount"] < Q1 - 1.5 * IQR) |
    (banking["Transaction_Amount"] > Q3 + 1.5 * IQR)
]

print("Transaction Amount Outliers:", len(outliers))

# Account_Balance

Q1 = banking["Account_Balance"].quantile(0.25)
Q3 = banking["Account_Balance"].quantile(0.75)
IQR = Q3 - Q1

outliers = banking[
    (banking["Account_Balance"] < Q1 - 1.5 * IQR) |
    (banking["Account_Balance"] > Q3 + 1.5 * IQR)
]

print("Account Balance Outliers:", len(outliers))
```

```python
# Which day of the week has the highest fraud?
banking["Day_Name"] = banking["Transaction_Date"].dt.day_name()
print(banking[["Transaction_Date", "Day_Name"]].head(10))

fraud = banking[banking["Is_Fraud"] == 1]
print(fraud["Day_Name"].value_counts())
print(fraud["Day_Name"].value_counts().idxmax())
print(fraud["Day_Name"].value_counts(normalize=True) * 100)
```

```python
#Financial Analysis
```

```python
#Are high-value transactions more likely to be fraudulent?

import numpy as np
conditions = [
    banking["Transaction_Amount"] < 25000,
    banking["Transaction_Amount"].between(25000, 50000),
    banking["Transaction_Amount"].between(50001, 75000),
    banking["Transaction_Amount"] > 75000
]
choices = [
    "Low",
    "Medium",
    "High",
    "Very High"
]
banking["Transaction_Size"] = np.select(conditions, choices, default="Unknown")
print(banking[["Transaction_Amount", "Transaction_Size"]].head(10))
fraud = banking[banking["Is_Fraud"] == 1]
print(fraud["Transaction_Size"].value_counts())
print(fraud["Transaction_Size"].value_counts().idxmax())
print(fraud["Transaction_Size"].value_counts(normalize=True)*100)
```

```python
# Does account balance affect fraud?
conditions=[
    banking["Account_Balance"]<25000,
    banking["Account_Balance"].between(25000,50000),
    banking["Account_Balance"].between(50001,75000),
    banking["Account_Balance"]>75000
]
choices=["Low",
         "Medium",
         "High",
         "Very High"
         ]
banking["Balance"]=np.select(conditions,choices,default="Unknown")
print(banking[["Account_Balance","Balance"]].head(10))
fraud = banking[banking["Is_Fraud"] == 1]
print(fraud["Balance"].value_counts())
print(fraud["Balance"].value_counts().idxmax())
print(fraud["Balance"].value_counts(normalize=True) * 100)
```

```python
# Which merchant category has the highest average fraudulent transaction amount?
fraud = banking[banking["Is_Fraud"] == 1]
average_amount = fraud.groupby("Merchant_Category")["Transaction_Amount"].mean()
print(average_amount)
print(average_amount.idxmax())
print(average_amount.max())
print(average_amount.sort_values(ascending=False))
```
